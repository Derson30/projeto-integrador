```python
import pandas as pd
import matplotlib.pyplot as plt

# Data provided by the user
days = ["Seg", "Ter", "Qua", "Qui", "Sex", "Sab", "Dom"]
average_temperatures = [24.5, 25.1, 23.8, 26.3, 27.0, 28.2, 26.8]

# Create a DataFrame
data = pd.DataFrame({
    'Day': days,
    'Average Temperature': average_temperatures
})

# Create a line plot based on 'Storytelling with Data' principles
fig, ax = plt.subplots(figsize=(10, 6))

# Plot the line
ax.plot(data['Day'], data['Average Temperature'], marker='o', linestyle='-', color='#1f77b4', linewidth=2)

# Add direct labels to the data points
for i, temp in enumerate(data['Average Temperature']):
    ax.text(i, temp + 0.3, f'{temp:.1f}°C', ha='center', va='bottom', fontsize=10, color='black')

# Remove chart junk:
# Make grid lines subtle and only on the y-axis
ax.grid(axis='y', linestyle='--', alpha=0.6, color='gray')

# Remove top and right spines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Set a more descriptive title and potentially a subtitle
ax.set_title('Variação da Temperatura Média Durante a Semana', fontsize=16, loc='left', pad=20)
ax.text(0, 1.05, 'Temperaturas em Celsius', transform=ax.transAxes, fontsize=12, va='bottom', ha='left', color='gray')

# Improve axis labels
ax.set_xlabel('Dia da Semana', fontsize=12)
ax.set_ylabel('Temperatura Média (°C)', fontsize=12)

# Ensure y-axis starts slightly below the minimum value for better visualization
min_temp = min(average_temperatures)
max_temp = max(average_temperatures)
ax.set_ylim(min_temp - 1, max_temp + 1)

plt.tight_layout()
plt.show()

display(data)
```

This code first imports the necessary libraries, `pandas` for data handling and `matplotlib.pyplot` for plotting. It then creates a DataFrame from the provided `days` and `average_temperatures` data. Finally, it generates a line plot to visualize the average temperature trend over the week, including a title, axis labels, and a grid for better readability. The DataFrame is also displayed.
