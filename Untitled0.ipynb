```python
import pandas as pd
import matplotlib.pyplot as plt

# Data provided by the user
days = ["Seg", "Ter", "Qua", "Qui", "Sex", "Sab", "Dom"]
average_temperatures = [24.5, 25.1, 23.8, 26.3, 27.0, 28.2, 26.8]

# Create a DataFrame
data = pd.DataFrame({
    'Day': days,
    'Average Temperature': average_temperatures
})

# Create a line plot
plt.figure(figsize=(10, 6))
plt.plot(data['Day'], data['Average Temperature'], marker='o', linestyle='-', color='b')
plt.title('Average Temperature Over the Week')
plt.xlabel('Day of the Week')
plt.ylabel('Average Temperature (°C)')
plt.grid(True)
plt.show()

display(data)
```

This code first imports the necessary libraries, `pandas` for data handling and `matplotlib.pyplot` for plotting. It then creates a DataFrame from the provided `days` and `average_temperatures` data. Finally, it generates a line plot to visualize the average temperature trend over the week, including a title, axis labels, and a grid for better readability. The DataFrame is also displayed.
